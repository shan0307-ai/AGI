```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import TextLoader
import os

project_path = "/content/drive/MyDrive/rag_project/documents"

documents = []
for filename in os.listdir(project_path):
    loader = TextLoader(f"{project_path}/{filename}")
    documents.extend(loader.load())

print(f"Total documents loaded: {len(documents)}")

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=400,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(documents)

print(f"Total chunks created: {len(chunks)}")
print("\n--- Sample chunk ---")
print(chunks[0].page_content)
print("\nSource:", chunks[0].metadata)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

vectordb = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    persist_directory="/content/drive/MyDrive/rag_project/chroma_db"
)

print("Vector database created and saved with", vectordb._collection.count(), "chunks")
```

```python
def rag_answer(question, k=3):
    # Retrieve relevant chunks
    results = vectordb.similarity_search(question, k=k)
    context = "\n\n".join([doc.page_content for doc in results])

    # Build the prompt
    prompt = f"""Answer the question using ONLY the context below. If the answer is not in the context, say "I don't have that information."

Context:
{context}

Question: {question}

Answer:"""

    messages = [{"role": "user", "content": prompt}]
    output = pipe(messages, max_new_tokens=200)
    answer = output[0]["generated_text"][-1]["content"]

    return answer, results

# Test it
question = "How many leave days do I get?"
answer, sources = rag_answer(question)

print("QUESTION:", question)
print("\nANSWER:", answer)
print("\n--- Sources used ---")
for doc in sources:
    print("-", doc.metadata["source"])
```

```python
%%writefile app.py
import streamlit as st
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
import torch

st.set_page_config(page_title="Knowledge Assistant", page_icon="◆", layout="centered")

# ---------- Styling ----------
st.markdown("""
<style>
    .stApp { background-color: #F7F8FA; }
    #MainMenu, footer, header { visibility: hidden; }

    .app-header {
        padding: 1.75rem 0 1.25rem 0;
        border-bottom: 1px solid #E3E6EA;
        margin-bottom: 1.5rem;
    }
    .app-header h1 {
        font-size: 1.6rem;
        font-weight: 650;
        color: #1F2937;
        margin: 0;
        letter-spacing: -0.01em;
    }
    .app-header p {
        font-size: 0.92rem;
        color: #6B7280;
        margin: 0.3rem 0 0 0;
    }

    [data-testid="stChatMessage"] {
        background-color: #FFFFFF;
        border: 1px solid #E3E6EA;
        border-radius: 10px;
        padding: 0.4rem 0.2rem;
        margin-bottom: 0.6rem;
    }

    .stChatInput textarea {
        border-radius: 10px !important;
        border: 1px solid #D1D5DB !important;
    }

    section[data-testid="stSidebar"] {
        background-color: #1F2937;
    }
    section[data-testid="stSidebar"] * {
        color: #F3F4F6 !important;
    }
    .sidebar-tag {
        display: inline-block;
        background-color: #374151;
        border-radius: 6px;
        padding: 0.3rem 0.6rem;
        margin: 0.2rem 0.3rem 0.2rem 0;
        font-size: 0.85rem;
        border-left: 3px solid #D97706;
    }

    .source-tag {
        display: inline-block;
        background-color: #FFF7ED;
        color: #9A3412;
        border: 1px solid #FDBA74;
        border-radius: 6px;
        padding: 0.15rem 0.5rem;
        font-size: 0.78rem;
        margin: 0.15rem 0.3rem 0 0;
    }
</style>
""", unsafe_allow_html=True)

# ---------- Header ----------
st.markdown("""
<div class="app-header">
    <h1>Internal Knowledge Assistant</h1>
    <p>Ask about leave, IT support, expenses, or onboarding. Answers are grounded in company documents.</p>
</div>
""", unsafe_allow_html=True)

# ---------- Sidebar ----------
with st.sidebar:
    st.markdown("### What I can help with")
    st.markdown("""
    <span class="sidebar-tag">Leave & vacation</span>
    <span class="sidebar-tag">Work from home</span>
    <span class="sidebar-tag">Expense claims</span>
    <span class="sidebar-tag">IT support</span>
    <span class="sidebar-tag">Onboarding</span>
    """, unsafe_allow_html=True)
    st.markdown("---")
    st.markdown("Answers are generated by a small language model (Phi-3-mini) grounded in retrieved company documents. If something isn't covered, I'll say so rather than guess.")

# ---------- Backend ----------
@st.cache_resource
def load_pipeline():
    model_name = "microsoft/Phi-3-mini-4k-instruct"
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto", torch_dtype=torch.bfloat16)
    return pipeline("text-generation", model=model, tokenizer=tokenizer, max_new_tokens=150)

@st.cache_resource
def load_vectordb():
    embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
    return Chroma(persist_directory="/content/drive/MyDrive/rag_project/chroma_db", embedding_function=embedding_model)

pipe = load_pipeline()
vectordb = load_vectordb()

def rag_answer(question, k=2, threshold=0.15):
    results_with_scores = vectordb.similarity_search_with_relevance_scores(question, k=k)
    relevant_results = [(doc, score) for doc, score in results_with_scores if score >= threshold]

    if not relevant_results:
        return "I don't have that information in the available documents.", []

    context = "\n\n".join([doc.page_content for doc, score in relevant_results])
    results = [doc for doc, score in relevant_results]

    prompt = f"""You are a helpful HR assistant. Answer the employee's question using only the context below. State each relevant number individually; do not add or combine numbers from different categories.

Context:
{context}

Question: {question}

Answer:"""

    messages = [{"role": "user", "content": prompt}]
    output = pipe(messages)
    answer = output[0]["generated_text"][-1]["content"]
    return answer, results

# ---------- Chat ----------
if "messages" not in st.session_state:
    st.session_state.messages = []

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.write(msg["content"])
        if msg.get("sources"):
            tags = "".join(f'<span class="source-tag">{s}</span>' for s in msg["sources"])
            st.markdown(tags, unsafe_allow_html=True)

if question := st.chat_input("Ask a question..."):
    st.session_state.messages.append({"role": "user", "content": question})
    with st.chat_message("user"):
        st.write(question)

    with st.chat_message("assistant"):
        with st.spinner("Checking documents..."):
            answer, sources = rag_answer(question)
            st.write(answer)
            source_names = sorted(set(s.metadata["source"].split("/")[-1] for s in sources))
            if source_names:
                tags = "".join(f'<span class="source-tag">{s}</span>' for s in source_names)
                st.markdown(tags, unsafe_allow_html=True)

    st.session_state.messages.append({"role": "assistant", "content": answer, "sources": source_names if sources else []})
```
